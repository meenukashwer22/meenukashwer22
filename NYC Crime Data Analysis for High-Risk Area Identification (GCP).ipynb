```
# NYC Crime Data Analysis - High-Risk Area Identification
# CSIS 586 - Cloud and Big Data Analytics

# Step 1: Install necessary packages (uncomment if not installed)
# !pip install pandas plotly geopandas

# Step 2: Import libraries
import pandas as pd
import plotly.express as px
import geopandas as gpd

# Step 3: Load dataset
df = pd.read_csv("NYC_Crime_Data.csv")

# Step 4: Quick data overview
print("Columns:", df.columns)
print(df.head())
print("Missing values:\n", df.isnull().sum())

# Step 5: Data Cleaning
# Drop rows with missing coordinates
df = df.dropna(subset=['LAT', 'LON'])

# Convert date column to datetime
df['Date'] = pd.to_datetime(df['DATE OCC'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')
df = df.dropna(subset=['Date'])

# Step 6: Feature Engineering
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['DayOfWeek'] = df['Date'].dt.day_name()

# Step 7: Exploratory Analysis
# Top 10 Crime Types
top_crimes = df['Crm Cd Desc'].value_counts().nlargest(10)
print("Top 10 Crime Types:\n", top_crimes)

# Crimes by Borough
borough_counts = df['AREA NAME'].value_counts()
print("Crimes by Borough:\n", borough_counts)

# Step 8: Spatial Visualization - NYC Crime Map
fig_map = px.scatter_mapbox(
    df.sample(5000),  # Plot a sample for performance
    lat="LAT",
    lon="LON",
    color="Crm Cd Desc",
    hover_data=["AREA NAME", "Date"],
    zoom=10,
    height=600
)
fig_map.update_layout(mapbox_style="open-street-map")
fig_map.update_layout(title="NYC Crime Locations")
fig_map.show()

# Step 9: Time-Series Analysis - Crimes Over Time
crimes_over_time = df.groupby('Year').size().reset_index(name='Count')
fig_time = px.line(
    crimes_over_time,
    x='Year',
    y='Count',
    title='NYC Crime Trends Over Years'
)
fig_time.show()

# Step 10: Heatmap by Borough and Crime Type
heatmap_data = df.groupby(['AREA NAME', 'Crm Cd Desc']).size().reset_index(name='Count')
fig_heat = px.density_heatmap(
    heatmap_data,
    x='AREA NAME',
    y='Crm Cd Desc',
    z='Count',
    color_continuous_scale='Viridis',
    title='Crime Density by Borough and Type'
)
fig_heat.show()
```
