```
# Step 1: Import libraries
import pandas as pd
import plotly.express as px
from jupyter_dash import JupyterDash
from dash import dcc, html
import folium
from folium.plugins import HeatMap

# Step 2: Load dataset
df = pd.read_csv("NYC_Crime_Data.csv")

# Step 3: Clean and preprocess
# Standardize column names
df.columns = df.columns.str.strip().str.upper().str.replace(' ', '_')

# Set column names for our use
date_col = 'DATE_OCC'       # Date of occurrence
lat_col = 'LAT'
lon_col = 'LON'
crime_col = 'CRM_CD_DESC'
area_col = 'AREA_NAME'

# Convert date column to datetime
df[date_col] = pd.to_datetime(df[date_col], errors='coerce')

# Extract hour and day of week
df['HOUR'] = df[date_col].dt.hour
df['DAY_OF_WEEK'] = df[date_col].dt.day_name()

# Drop rows with missing lat/lon
df_clean = df.dropna(subset=[lat_col, lon_col])

# Optional: Filter for recent years
df_clean = df_clean[df_clean[date_col].dt.year >= 2015]

# -------------------------------
# Step 4: Prepare Plotly charts
# -------------------------------

# 4a: Crimes by Area (borough equivalent)
borough_counts = df_clean[area_col].value_counts().reset_index()
borough_counts.columns = ['Borough', 'Crime Count']
fig_borough = px.bar(
    borough_counts,
    x='Borough',
    y='Crime Count',
    color='Crime Count',
    title='Total Crimes by Area'
)

# 4b: Crime by Hour
hour_counts = df_clean.groupby('HOUR').size().reset_index(name='Crime Count')
fig_hour = px.line(
    hour_counts,
    x='HOUR',
    y='Crime Count',
    markers=True,
    title='Crime Trend by Hour'
)

# 4c: Crime by Day of Week
day_counts = df_clean['DAY_OF_WEEK'].value_counts().reindex(
    ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
).reset_index()
day_counts.columns = ['DayOfWeek', 'Crime Count']
fig_day = px.bar(
    day_counts,
    x='DayOfWeek',
    y='Crime Count',
    color='Crime Count',
    title='Crimes by Day of the Week'
)

# -------------------------------
# 4d: NYC Crime Heatmap with Folium
# -------------------------------
heat_data = df_clean[[lat_col, lon_col]].values.tolist()
nyc_map = folium.Map(location=[40.7128, -74.0060], zoom_start=11)
HeatMap(heat_data, radius=10).add_to(nyc_map)
nyc_map.save("nyc_crime_heatmap.html")

print("Heatmap saved as 'nyc_crime_heatmap.html'")
```

```
# NYC Crime Data Analysis - High-Risk Area Identification
# CSIS 586 - Cloud and Big Data Analytics

# Step 1: Install necessary packages (uncomment if not installed)
# !pip install pandas plotly geopandas

# Step 2: Import libraries
import pandas as pd
import plotly.express as px
import geopandas as gpd

# Step 3: Load dataset
df = pd.read_csv("NYC_Crime_Data.csv")

# Step 4: Quick data overview
print("Columns:", df.columns)
print(df.head())
print("Missing values:\n", df.isnull().sum())

# Step 5: Data Cleaning
# Drop rows with missing coordinates
df = df.dropna(subset=['LAT', 'LON'])

# Convert date column to datetime
df['Date'] = pd.to_datetime(df['DATE OCC'], errors='coerce')
df = df.dropna(subset=['Date'])

# Step 6: Feature Engineering
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['DayOfWeek'] = df['Date'].dt.day_name()

# Step 7: Exploratory Analysis
# Top 10 Crime Types
top_crimes = df['Crm Cd Desc'].value_counts().nlargest(10)
print("Top 10 Crime Types:\n", top_crimes)

# Crimes by Borough
borough_counts = df['AREA NAME'].value_counts()
print("Crimes by Borough:\n", borough_counts)

# Step 8: Spatial Visualization - NYC Crime Map
fig_map = px.scatter_mapbox(
    df.sample(5000),  # Plot a sample for performance
    lat="LAT",
    lon="LON",
    color="Crm Cd Desc",
    hover_data=["AREA NAME", "Date"],
    zoom=10,
    height=600
)
fig_map.update_layout(mapbox_style="open-street-map")
fig_map.update_layout(title="NYC Crime Locations")
fig_map.show()

# Step 9: Time-Series Analysis - Crimes Over Time
crimes_over_time = df.groupby('Year').size().reset_index(name='Count')
fig_time = px.line(
    crimes_over_time,
    x='Year',
    y='Count',
    title='NYC Crime Trends Over Years'
)
fig_time.show()

# Step 10: Heatmap by Borough and Crime Type
heatmap_data = df.groupby(['AREA NAME', 'Crm Cd Desc']).size().reset_index(name='Count')
fig_heat = px.density_heatmap(
    heatmap_data,
    x='AREA NAME',
    y='Crm Cd Desc',
    z='Count',
    color_continuous_scale='Viridis',
    title='Crime Density by Borough and Type'
)
fig_heat.show()
```
