```
#Install packages (in Colab or local Python)

import requests
import pandas as pd

#Fetch Data from API

url = "https://www.alphavantage.co/query?function=TIME_SERIES_DAILY&symbol=AAPL&apikey=8XSP9CKYX3VFQBYS"

response = requests.get(url)
data = response.json()

#Convert JSON → DataFrame

time_series = data["Time Series (Daily)"]

df = pd.DataFrame.from_dict(time_series, orient="index")

df = df.rename(columns={
    "1. open": "Open",
    "2. high": "High",
    "3. low": "Low",
    "4. close": "Close",
    "5. volume": "Volume"
})

df.index = pd.to_datetime(df.index)
df.reset_index(inplace=True)
df = df.rename(columns={"index": "Date"})

#Save as CSV

df.to_csv("stock_data.csv", index=False)
```
